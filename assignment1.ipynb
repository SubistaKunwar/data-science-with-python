This notebook contains my Pandas assignment. I used Pandas to read and clean the order dataset, handle missing values, remove duplicate orders, filter records, and create new columns.

## Step 1: Read the CSV File

Load the CSV file into a Pandas DataFrame and display the first 5 rows.

```python
import pandas as pd

df = pd.read_csv('pandas_dataset.csv')
before_shape = df.shape

print('First 5 rows:')
print(df.head())
```

## Step 2: Rename Columns

Remove extra spaces, convert column names to lowercase, and replace spaces with underscores.

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
)

print('Clean column names:')
print(df.columns.tolist())
```

## Step 3: Remove Extra Empty Rows

The exported CSV contains a completely empty row. I removed rows where all values are missing and reset the index.

```python
print('Before removing empty rows:')
print(df.head())

df = df.dropna(how='all')
df = df.reset_index(drop=True)

print('\nAfter removing empty rows:')
print(df.head())
```

## Step 4: Drop Unnecessary Columns

I checked the columns and did not drop any because each column provides useful information about the order, customer, product, sales, profit, or discount.

```python
print('Columns in the dataset:')
print(df.columns.tolist())

print('\nNo columns were dropped.')
print('Reason: All columns contain useful information about the orders.')
```

## Step 5: Count Missing Values

Count the null values in each column.

```python
missing_values = df.isnull().sum()

print('Missing values in each column:')
print(missing_values)
```

## Step 6: Handle Missing Values

- Drop rows where `sales` is missing.
- Fill missing `profit` values with the mean.
- Fill missing `discount` values with the median.

```python
# 6(a): Drop rows where sales is missing
df = df.dropna(subset=['sales'])

# 6(b): Fill missing profit with mean
profit_mean = df['profit'].mean()
df['profit'] = df['profit'].fillna(profit_mean)

# Fill missing discount with median
discount_median = df['discount'].median()
df['discount'] = df['discount'].fillna(discount_median)

print('Mean profit used:', profit_mean)
print('Median discount used:', discount_median)
print('\nMissing values after Step 6:')
print(df[['sales', 'profit', 'discount']].isnull().sum())
```

## Step 7: Handle Missing Customer Names

Replace missing customer names with `Unknown`.

```python
df['customer_name'] = df['customer_name'].fillna('Unknown')

print('Customer names after handling missing values:')
print(df['customer_name'])
```

## Step 8: Detect and Remove Duplicates

Check for duplicate records based on `order_id`, display them, count them, and remove the duplicate rows.

```python
duplicates = df.duplicated(subset=['order_id'])

print('Duplicate rows:')
print(df[duplicates])

duplicate_count = duplicates.sum()
print('\nNumber of duplicate rows removed:', duplicate_count)

df = df.drop_duplicates(subset=['order_id'], keep='first')
df = df.reset_index(drop=True)
```

## Step 9: Filtering and Creating New Columns

### 9(a) Orders where `unit_price > 20,000`

### 9(b) Orders where `unit_price > 10,000` and `status == Completed`

### 9(c) Create `total_amount`

`total_amount = quantity × unit_price − discount`

### 9(d) Create `customer_type`

If quantity is 3 or more, the customer is a `Bulk Buyer`; otherwise, `Regular Buyer`.

```python
# 9(a): Orders where unit_price > 20,000
print('Orders where unit_price > 20,000:')
orders_above_20000 = df[df['unit_price'] > 20000]
print(orders_above_20000)

# 9(b): unit_price > 10,000 AND status == Completed
print('\nCompleted orders with unit_price > 10,000:')
completed_orders = df[
    (df['unit_price'] > 10000) &
    (df['status'] == 'Completed')
]
print(completed_orders[['customer_name', 'category', 'status']])

# 9(c): Create total_amount
df['total_amount'] = (
    df['quantity'] * df['unit_price'] - df['discount']
)

print('\nDataset with total_amount:')
print(df[['quantity', 'unit_price', 'discount', 'total_amount']])

# 9(d): Create customer_type
df['customer_type'] = df['quantity'].apply(
    lambda x: 'Bulk Buyer' if x >= 3 else 'Regular Buyer'
)

print('\nDataset with customer_type:')
print(df[['customer_name', 'quantity', 'customer_type']])
```

## Step 10: Final Cleaned Dataset

Display the shape before and after cleaning and show the first 10 rows of the final dataset.

```python
after_shape = df.shape

print('Shape before cleaning:')
print(before_shape)

print('\nShape after cleaning:')
print(after_shape)

print('\nFirst 10 rows of cleaned dataset:')
print(df.head(10))
```
